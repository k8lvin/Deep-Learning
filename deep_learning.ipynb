## The parts of a neural network    
### 1. Neuron(Units)     
A neuron takes a set of inputs, multiplies each by a weight, sums them together with a bias term, and passes through an activation function. 

_Single neuron example_    
$y = \lambda(\sum_{i=1}^{n} w_i x_i + b)$     

y = Output     
$\lambda$ = Activation function      
$\sum_{i=1}^{n}$ = Summation     
$w_i$ = Weight of input i     
$x_i$ = Input i              
b = bias      

_expanded example_     
$y = \lambda(w_1 x_1  + w_2 x_2  + w_3 x_3 + b)$     

### 2. weights and biases        
- __Weights:__ Control how strongly each input influences a neuron's output. Learning a network mostly means adjusting these weights.         
- __Bias:__ An extra learnable value added to the weighted sum, letting the neuron shift its output up or down independent of the input.     

### 3. layers    
- __Input layer:__ Recieves the raw data (e.g pixel values, numeric features)      
- __Hidden layer(s):__ Intermediate layers that transform the data.  "Deep" learning simply means having many hidden layers.       
- __Output layer:__ Produces the final prediction(e.g a class probability or a numeric value).      

### 4. Forward propagation      
Forward propagation is the process of passing an input through a the network, layer by layer, until it produces an output/prediction. Each layer'a output becomes the next layer's input.     

$y = \lambda(w_1 x_1  + w_2 x_2  + w_3 x_3.....w_{142191} x_{142191} + b)$     

### 5. Loss Function/Cost function     
The loss function measures how far the network's prediction is from the true answer.      
Common choices include _mean squared error_ for regression problems and _cross entropy_ for classification problems.       
Training tries to minimize this loss.     

### 6. Backpropagation and Optimizers      
__Backpropagation__ is the algorithm that computes how much each weight contributes to the error, working backward from the output layer to the input layer.   

__Optimers__ e.g Adam, Stochastic Gradient Descent. uses the computed gradients to update each weight, nudging it in the direction that reduces the loss.  

__Learning rate:__ How big a step the optimizer takes when updating weights.      
__Epoch:__ One complete pass of the entire training dataset through the model.      
__Batch Size:__ The number of training examples processed before the weights are updated once.     

Over many epochs, the loss function should trend downward as the network improves.     

### Activation Functions in Neural Networks    
Without an activation funtion, a neural network would just be a stack of linear operations, which collapses into a single linear function no matteer how many layers you add.        
Activation functions introduce non-linearity, which is what allows neural networks to model complex, curved decision boundaries and relationships.    

### 1. Sigmoid     
_Formula_     
$\sigma(x) = \frac{1}{1 + e^{-x}}$      

- Output range for sigmoid is a __Single__ value between 0 and 1        
- Squashes any input into a probability-like range.       
- Commonly used today mainly in the output layer for binary classification.     
- Suffers from vanishing gradient for very small or very large inputs, which can slow down learning in deep networks.    

Gradient for point(1, -2)     

Point A(0, -2)

### 2. Tanh(Hyperbolic tangent)     
_Formula_     
$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$     

- It has an output range of (-1, 1)     
- Zero centered        
- Still suffers vanishing gradient at the extremes.     

### 3. SoftMax     
_Formula_     
$\sigma (z)_{i}=\frac{e^{z_{i}}}{\sum _{j=1}^{K}e^{z_{j}}}$    

The output range for softmax are values between 0 and 1, all outputs sum to 1.        
- Converts a vector of raw scores into probability distribution across multiple classes.    
- Used almost exclusively in the output layer for multiclass classification problems.      

### 4. ReLU(Rectified Linear Unit)    
_formula_     
$f(x) = max(0, x)$     

Output range of ReLU is 0 to infinity.     
- The most widely used activation function in hidden layers, helps reduce the vanishing gradient problem for positive values.      
- Can suffer from the "dying ReLU" problem, where neurons get stuck always outputting zero.    

### 5. Leaky ReLU    
_Formula_    
$f(x) = max(\alpha{x}, x)$       

The Output range for Leaky ReLU is between negative infinity and infinity     
A small modification of ReLU that allows a small, non-zero gradient when the input is negative.    
Help mitigate the dying ReLU problem.
